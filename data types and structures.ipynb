DATA TYPES AND STRUCTURES

#1 What are data structures, and why are they important?
   A data structure is a way of organizing and storing data in a computer so it can be accessed and used efficiently. Choosing the right data structure allows programmers to handle large amounts of data quickly, which is important for building fast and reliable computer programs.

#2 Explainthe difference between mutable and immutable data types with        examples?
   Mutable data types can be changed after creation (e.g., list, dictionary), while immutable data types cannot be altered once set (e.g., string, tuple). For example, a list () is mutable, but a string ("abc") is immutable.

#3 What are the main differences between lists and tuples in Python?
   Lists are mutable and can be changed after creation, while tuples are immutable and cannot be modified. Lists use square brackets []; tuples use parentheses ()

#4 Describe how dictionaries store data?
   Dictionaries store data as key-value pairs, where each unique key maps to a specific value and values are accessed using their unique keys. Internally, dictionaries use a hashing mechanism to allow for efficient retrieval of values by their keys.

#5 Why might you use a set instead of a list in Python?
   A set in Python is used instead of a list when only unique elements are needed and fast membership testing (checking if an item exists) is important, since sets automatically remove duplicates and are much faster for these operations than lists. Lists should be used when order matters or duplicate items are required.

#6 What is a string in Python, and how is it different from a list?
   A string in Python is an immutable sequence of characters enclosed in quotes, while a list is a mutable sequence of items that can be of any data type. Unlike strings, lists can be changed after creation by adding, removing, or modifying elements.

#7 How do tuples ensure data integrity in Python?
   Tuples ensure data integrity in Python by being immutable, which means once created, their contents cannot be changed, preventing accidental or unauthorized modifications. This immutability guarantees that the data remains consistent and reliable throughout the program's execution.

#8 What is a hash table, and how does it relate to dictionaries in Python?
   A hash table is a data structure that stores key-value pairs and uses a hash function to compute an index for fast data retrieval. Python dictionaries are implemented as hash tables, enabling efficient lookup, insertion, and deletion of key-value pairs.

#9 Can lists contain different data types in Python?
   Yes, Python lists can contain different data types simultaneously, such as integers, strings, and booleans in the same list. This flexibility makes lists very versatile for storing heterogeneous data items.

#10 Explain why strings are immutable in Python?
     Strings in Python are immutable to ensure safety from unintended modifications, enabling reliable hashing for use as dictionary keys and improving performance optimization. This design avoids bugs by preventing changes to string objects once created.
   
#11 What advantages do dictionaries offer over lists for certain tasks ?
  Dictionaries offer faster lookup and access by keys with average time complexity O(1), unlike lists which require O(n) time for searching. They also store data in key-value pairs, providing more meaningful and flexible data representation compared to the indexed elements of lists.

#12 Describe a scenario where using a tuple would be preferable over a list ?
    
  Tuples are preferable when you need an immutable, fixed collection of items that won't change, offering better performance and memory efficiency than lists. They are ideal for storing constant data like coordinates or keys in dictionaries.

#13 How do sets handle duplicate values in Python?
   Python sets automatically remove duplicate values, ensuring only one copy of each element is stored. When duplicates are added, they are simply ignored, maintaining uniqueness within the set.

#14 How does the “in” keyword work differently for lists and dictionaries?
  The "in" keyword checks for membership in lists by scanning each element until a match is found (O(n) time), while for dictionaries it checks for the presence of a key using a hash lookup, making it much faster (O(1) average time complexity).

#15 Can you modify the elements of a tuple? Explain why or why not?
  You cannot modify the elements of a tuple because tuples in Python are immutable, meaning their contents cannot be changed once created. However, you can work around this by converting the tuple to a list, modifying the list, and then converting it back to a tuple.

#16  What is a nested dictionary, and give an example of its use case?
  A nested dictionary is a dictionary in which the values are themselves dictionaries, allowing for hierarchical or grouped data storage. For example, storing student information with each student ID as a key and their details like name, age, and grade in an inner dictionary is a common use case.

#17 Describe the time complexity of accessing elements in a dictionary?
   The time complexity of accessing elements in a dictionary by key is on average O(1) due to the hash table implementation, allowing direct retrieval. However, in the worst case, it can degrade to O(n) if many hash collisions occur. This makes dictionary key access highly efficient for most practical use cases.

#18 In what situations are lists preferred over dictionaries?
   Lists are preferred over dictionaries when the order of elements matters, and when the collection needs to be mutable and sequentially accessed, such as for storing ordered data or maintaining sequences of items. Lists are also more memory-efficient and suitable for cases where quick key-based lookups are not required.

#19 Why are dictionaries considered unordered, and how does that affect data retrieval?
   Dictionaries are considered unordered because their primary structure is a hash table optimized for fast key-based access rather than maintaining items in a specific sequence. This unordered nature means data retrieval is done by keys without relying on the order of insertion, so you cannot access items by index as you would in ordered collections like lists.

#20  Explain the difference between a list and a dictionary in terms of data retrieval?
   A list retrieves data by accessing elements through integer indices in a specific order, while a dictionary retrieves data using unique keys with average constant time complexity due to hash mapping. Lists maintain order and allow duplicates, whereas dictionaries map keys to values with unique keys and focus on fast lookups.
